```python
from backtesting import Strategy

class ImprovedStrategy2(Strategy): # 版本二(修正架構)
    def init(self):
        # 初始化策略
        self.i = 0
        self.count1 = 0
        self.count2 = 0
        self.lessamount = 0
        self.higher_high = 0
        self.higher_low = 0
        self.lower_high = 0
        self.lower_low = 0

    def next(self):
        # 定義策略的執行邏輯
          current_high = self.data.High[self.i]
          current_low = self.data.Low[self.i]

          if self.i < 3000000:

            self.i += 1

            if current_high > self.higher_high and current_low > self.higher_low:
                self.count1 += 1

                if self.count1 - self.count2 >= 80: # 如果連續出現更高高點和更高低點，執行賣出操作
                    sell_amount = 10000000

                    if self.lessamount >= 1 and self.lessamount < 20:  # 賣出資金
                      self.sell(size=sell_amount * self.lessamount)
                      self.lessamount = 0

                    self.higher_high = current_high
                    self.higher_low = current_low
                    self.lower_high = current_high
                    self.lower_low = current_low

                    self.count1 = 0
                    self.count2 = 0


            elif current_high < self.lower_high and current_low < self.lower_low:
                self.count2 += 1

                if self.count2 - self.count1 >= 40: # 如果連續出现更低高點和更低低點，執行買入操作
                  buy_amount = 10000000 # 買入资金
                  self.lessamount += 1

                  if self.lessamount < 10:
                    self.buy(size=buy_amount)  # 一次買入10000000元

                  self.higher_high = current_high
                  self.higher_low = current_low
                  self.lower_high = current_high
                  self.lower_low = current_low

                  self.count1 = 0
                  self.count2 = 0
```
